```python
import numpy as np
import pandas as pd

# Load processed data
X_train = np.load("../data/processed/X_train_scaled.npy")
X_valid = np.load("../data/processed/X_valid_scaled.npy")
y_train = np.load("../data/processed/y_train.npy")
y_valid = np.load("../data/processed/y_valid.npy")
feature_names = pd.read_csv("../data/processed/feature_names.csv", header=None)[0].tolist()
```

```python
# Import Random Forest classification model and evaluation metrics to assess the model

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, roc_auc_score, accuracy_score, confusion_matrix, log_loss, make_scorer

# Initialize the baseline Random Forest model with a defined random state for reproducible results. 

rf_baseline_model = RandomForestClassifier(random_state=42)

#  Fit the model to the training data

rf_baseline_model.fit(X_train, y_train)

print("Baseline Random Forest Model successfully initialized and trained on the training data.")
```

```python
# Import evaluation function that takes a specified model and test data as inputs, 
# uses the model to predict Gallstone Status and outputs evaluation metrics of the model.
# See code for evaluate_model function in evaluation_utility.py
 
from Utility_Functions.evaluation_utility import evaluate_model

evaluate_model(rf_baseline_model, X_valid, y_valid, model_name="Baseline Random Forest Model")
```

```python
# Import hyperparameter tuning and evaluation function from file
# See code for hyperparameter tuning function in model_tuning_utility.py

from Utility_Functions.model_tuning_utility import optimize_tree_model
from Utility_Functions.evaluation_utility import evaluate_model

# 1st iteration of Hyperparameter tuned model using 2 folds

rf_tuned_model_1 = optimize_tree_model('RF', n_splits=2, X_train=X_train, y_train=y_train, n_repeats=1, n_iter_search=60, scoring='balanced_accuracy')

evaluate_model(rf_tuned_model_1, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=2)")
```

```python
# 2nd iteration of Hyperparameter tuned model using 3 folds, scoring metric balanced_accuracy

rf_tuned_model_2 = optimize_tree_model('RF', n_splits=3, X_train=X_train, y_train=y_train, n_repeats=1, n_iter_search=60, scoring='balanced_accuracy')

evaluate_model(rf_tuned_model_2, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=3)")
```

```python
# 3rd iteration of Hyperparameter tuned model using 3 folds with 200 candidates

rf_tuned_model_3 = optimize_tree_model('RF', n_splits=3, X_train=X_train, y_train=y_train, n_repeats=1, n_iter_search=200, scoring='balanced_accuracy')

evaluate_model(rf_tuned_model_3, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=3)")
```

```python
# 4th iteration of Hyperparameter tuned model using 15 evaluations with 200 candidates

rf_tuned_model_4 = optimize_tree_model('RF', n_splits=3, X_train=X_train, y_train=y_train, n_repeats=5, n_iter_search=200, scoring='balanced_accuracy')

evaluate_model(rf_tuned_model_4, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=3, 5 repeats)")
```

```python
# 5th iteration of Hyperparameter tuned model using 9 evaluations with 200 candidates, auc scoring

rf_tuned_model_5 = optimize_tree_model('RF', n_splits=3, X_train=X_train, y_train=y_train, n_repeats=3, n_iter_search=200, scoring='auc')

evaluate_model(rf_tuned_model_5, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=3, 3 repeats)")
```

```python
# 6th iteration of Hyperparameter tuned model using 15 evaluations with 200 candidates, auc scoring

rf_tuned_model_6 = optimize_tree_model('RF', n_splits=5, X_train=X_train, y_train=y_train, n_repeats=3, n_iter_search=200, scoring='auc')

evaluate_model(rf_tuned_model_6, X_valid, y_valid, model_name="Hyperparameter tuned Random Forest Model (CV=5, 3 repeats)")
```

```python
import joblib
import os

# Saving the trained models to directory, results are presented in a separate notebook.

TARGET_DIR = r"C:\Python DAT540\Dat-540-Prosjekt\data\tree_models"

# Create the directory if it does not exist
if not os.path.exists(TARGET_DIR):
    os.makedirs(TARGET_DIR)
    print(f"Created directory: {TARGET_DIR}")

# Define the models list and a corresponding list of clean filenames
models_to_save = [
    rf_baseline_model, rf_tuned_model_1, rf_tuned_model_2, rf_tuned_model_3,
    rf_tuned_model_4, rf_tuned_model_5, rf_tuned_model_6
]

model_filenames = [
    'rf_baseline_model.joblib',
    'rf_tuned_model_1.joblib',
    'rf_tuned_model_2.joblib',
    'rf_tuned_model_3.joblib',
    'rf_tuned_model_4.joblib',
    'rf_tuned_model_5.joblib',
    'rf_tuned_model_6.joblib'
]

# Loop through and save each model
print("\nSaving models...")
for model, filename in zip(models_to_save, model_filenames):
    
    # os.path.join constructs the full, correct path: C:\...\tree_models\filename.joblib
    full_path = os.path.join(TARGET_DIR, filename) 
    
    joblib.dump(model, full_path)
    print(f"Saved: {full_path}")

print("\nAll models saved successfully.")
```
