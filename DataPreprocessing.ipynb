```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv('data/dataset-uci.csv')

# Step 1: Inspect the first few rows
print("First few rows of the dataset:")
print(df.head())
print("\n" + "="*80 + "\n")

# Step 2: Check datatypes and basic info
print("Dataset Information:")
print(df.info())
print("\n" + "="*80 + "\n")

# Step 3: Check dataset shape
print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
print("\n" + "="*80 + "\n")

# Step 4: Check for missing values
print("Missing values per column:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
print("\n" + "="*80 + "\n")

# Step 5: Statistical summary
print("Statistical Summary:")
print(df.describe())
```

### Encode categorical variables

```python
# Step 6: Encode categorical variables
df_encoded = df.copy()

# Identify categorical columns
categorical_cols = df_encoded.select_dtypes(include=['object']).columns
print(f"Categorical columns found: {list(categorical_cols)}")

# Create label encoders for each categorical column
label_encoders = {}
for col in categorical_cols:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col])
    label_encoders[col] = le
    print(f"\n{col} encoding:")
    print(f"  Original values: {df[col].unique()}")
    print(f"  Encoded values: {df_encoded[col].unique()}")

print("\n" + "="*80 + "\n")
print("Data after encoding:")
print(df_encoded.head())
```

### Normalize/Scale numeric features

```python
# Step 7: Normalize/Scale numeric features
# The first column is the target variable
feature_cols = df_encoded.columns[1:]  # All columns except the first one
target_col = df_encoded.columns[0]     # First column as target

print(f"Feature columns: {list(feature_cols)}")
print(f"Target column: {target_col}")

# Separate features and target
X = df_encoded[feature_cols]
y = df_encoded[target_col]

# Scale the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for better readability
X_scaled_df = pd.DataFrame(X_scaled, columns=feature_cols)

print("\n" + "="*80 + "\n")
print("Scaled features (first 5 rows):")
print(X_scaled_df.head())
print("\n" + "="*80 + "\n")
print("Scaling statistics:")
print(f"Mean: {X_scaled_df.mean().values}")
print(f"Std: {X_scaled_df.std().values}")
```

### Summary of preprocessing

```python
# Step 8: Summary of preprocessing
print("="*80)
print("PREPROCESSING SUMMARY")
print("="*80)
print(f"Original dataset shape: {df.shape}")
print(f"Preprocessed dataset shape: {X_scaled_df.shape}")
print(f"Number of features: {X_scaled_df.shape[1]}")
print(f"Number of samples: {X_scaled_df.shape[0]}")
print(f"Target variable unique values: {y.unique()}")
print(f"Target variable distribution:\n{y.value_counts()}")
```
